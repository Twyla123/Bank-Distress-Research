```python
import os
import pandas as pd
import zipfile
```

```python
def print_files_with_agency_code(summary_file):
    """
    Prints file names from `column_names_summary.csv` that include 
    `Agency Code` or `agency_code` in the `Column Name` column.

    Args:
        summary_file (str): Path to the `column_names_summary.csv` file.

    Returns:
        None
    """
    try:
        # Read the summary file
        df = pd.read_csv(summary_file)

        # Filter rows where 'Column Name' is exactly 'Agency Code' or 'agency_code'
        filtered_df = df[df['Column Name'].isin(['Agency Code', 'agency_code'])]

        # Extract unique file names
        file_names = filtered_df['File Name'].unique()

        # Print the filtered file names
        print("Files containing 'Agency Code' or 'agency_code' in the Column Name:")
        for file_name in file_names:
            print(f"  - {file_name}")

    except FileNotFoundError:
        print(f"Error: The file {summary_file} does not exist.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
summary_file_path = '/Users/twylazhang/Desktop/Bank_Research/HMDA/column_names_summary.csv'
print_files_with_agency_code(summary_file_path)
```

# directly drop 7 and 9 filtered LAR

```python
# Define the source and destination folder paths
source_folder_path = '/Users/twylazhang/Desktop/Bank_Research/HMDA/filtered_lar_data'
destination_folder_path = '/Users/twylazhang/Desktop/Bank_Research/HMDA/filtered_agency_code'

# Ensure the destination folder exists
os.makedirs(destination_folder_path, exist_ok=True)

# Possible column names for agency codes
agency_code_variations = ["Agency Code", "agency_code", "agency code"]

# Iterate over all CSV files in the source folder
for file_name in os.listdir(source_folder_path):
    file_path = os.path.join(source_folder_path, file_name)
    
    # Check if the file is a CSV file
    if file_name.endswith('.csv'):
        try:
            # Read the CSV file with low_memory=False to avoid dtype warnings
            df = pd.read_csv(file_path, low_memory=False, on_bad_lines='skip')
            
            # Check for the existence of agency code columns
            for col in agency_code_variations:
                if col in df.columns:
                    # Filter out rows where agency code is 7 or 9
                    df_filtered = df[~df[col].isin([7, 9])]
                    
                    # Extract the year from the file name  
                    year = ''.join(filter(str.isdigit, file_name))
                    
                    # Save the filtered data with the specified file name format
                    output_file_name = f"{year}_agency_filtered_lar.csv"
                    output_file_path = os.path.join(destination_folder_path, output_file_name)
                    df_filtered.to_csv(output_file_path, index=False)
                    
                    # Print confirmation and unique agency codes
                    unique_codes = df_filtered[col].dropna().unique()
                    print(f"Filtered data saved to {output_file_path}")
                    print(f"Unique agency codes for file {file_name} after filtering: {unique_codes}")
                    break  # Exit after finding the first matching column
        except Exception as e:
            print(f"Error processing file {file_name}: {e}")
```

```python
import os
import pandas as pd

# Define the source and destination folder paths
source_folder_path = '/Users/twylazhang/Desktop/Bank_Research/HMDA/merged_panel_ts_files'
destination_folder_path = '/Users/twylazhang/Desktop/Bank_Research/HMDA/filtered_agency_code'

# Ensure the destination folder exists
os.makedirs(destination_folder_path, exist_ok=True)

# Possible column names for agency codes
agency_code_variations = ["Agency Code", "agency_code", "agency code"]

# Iterate over all CSV files in the source folder
for file_name in os.listdir(source_folder_path):
    file_path = os.path.join(source_folder_path, file_name)
    
    # Check if the file is a CSV file
    if file_name.endswith('.csv'):
        try:
            # Read the CSV file with low_memory=False to avoid dtype warnings
            df = pd.read_csv(file_path, low_memory=False, on_bad_lines='skip')
            
            # Check for the existence of agency code columns
            for col in agency_code_variations:
                if col in df.columns:
                    # Filter out rows where agency code is 7 or 9
                    df_filtered = df[~df[col].isin([7, 9])]
                    
                    # Extract the year from the file name (assuming it's part of the file name)
                    year = ''.join(filter(str.isdigit, file_name))
                    
                    # Save the filtered data with the specified file name format
                    output_file_name = f"{year}_agency_filtered_mergedTS_Panel.csv"
                    output_file_path = os.path.join(destination_folder_path, output_file_name)
                    df_filtered.to_csv(output_file_path, index=False)
                    
                    # Print confirmation and unique agency codes
                    unique_codes = df_filtered[col].dropna().unique()
                    print(f"Filtered data saved to {output_file_path}")
                    print(f"Unique agency codes for file {file_name} after filtering: {unique_codes}")
                    break  # Exit after finding the first matching column
        except Exception as e:
            print(f"Error processing file {file_name}: {e}")
```
