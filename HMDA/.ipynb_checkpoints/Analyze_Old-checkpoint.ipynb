````python
import zipfile
import pandas as pd
import os
````

# compile the column names

````python
def extract_zip_and_expand_column_names_properly(output_file):
    """
    Extracts column names from CSV, Excel, and TXT files in ZIP archives in the current folder,
    and creates a structured output CSV where each column name is in its own row.

    Args:
        output_file (str): Path to save the output CSV file.
    """
    # Get the current working directory
    zip_folder = os.getcwd()
    data = []

    for zip_file in os.listdir(zip_folder):
        if zip_file.endswith('.zip'):
            # Extract year and file type
            if zip_file.startswith('hmda_'):
                # Handle 2011–2017 naming pattern
                year = next((part for part in zip_file.split('_') if part.isdigit()), "Unknown")
                file_type = zip_file.replace(f"hmda_{year}_", "").replace(".zip", "").replace('_', ' ').strip()
            else:
                # Handle other naming patterns
                parts = zip_file.split('_')
                year = next((part for part in parts if part.isdigit()), "Unknown")
                file_type = " ".join(part for part in parts if not part.isdigit()).replace('.zip', '').strip()

            # Open the ZIP file
            with zipfile.ZipFile(zip_file, 'r') as zip_ref:
                for file in zip_ref.namelist():
                    if file.endswith(('.csv', '.xlsx', '.xls', '.txt')):  # Check for supported formats
                        with zip_ref.open(file) as extracted_file:
                            try:
                                # Automatically detect delimiter for .csv and .txt
                                if file.endswith('.csv') or file.endswith('.txt'):
                                    sample = extracted_file.read(1024).decode('utf-8')  # Read a small sample to infer delimiter
                                    delimiter = ',' if ',' in sample else '|' if '|' in sample else '\t'  # Adjust as needed
                                    extracted_file.seek(0)  # Reset the file pointer
                                    df = pd.read_csv(extracted_file, delimiter=delimiter, nrows=0)
                                else:  # Excel files
                                    df = pd.read_excel(extracted_file, nrows=0)

                                # Collect each column name as a separate row
                                for column in df.columns:
                                    data.append({
                                        'Year': year,
                                        'File Type': file_type,
                                        'File Name': file,
                                        'Column Name': column.strip()  # Ensure clean column names
                                    })
                            except Exception as e:
                                print(f"Error processing {file} in {zip_file}: {e}")
                    else:
                        print(f"Ignored file: {file} in {zip_file} (unsupported format)")

    # Save the collected data to a CSV file
    df_output = pd.DataFrame(data)
    df_output.to_csv(output_file, index=False)
    print(f"Structured file with expanded column names saved to {output_file}")

# Example usage
output_csv_path = 'column_names_summary.csv'  # Output file name in the same directory
extract_zip_and_expand_column_names_properly(output_csv_path)
````

# Merge the Panel data with the TS data

````python
def merge_hmda_files(output_folder):
    """
    Merges the panel file with the transmittal sheet file for years 2011 to 2017,
    matching on Respondent ID. Adds "_p" suffix to all panel columns except the merging key
    and "_ts" suffix to all transmittal sheet columns. Handles 2016 TS data as a standalone CSV file.
    Saves the merged files in the specified output folder.

    Args:
        output_folder (str): Path to save the merged output files.
    """
    # Ensure the output folder exists
    os.makedirs(output_folder, exist_ok=True)

    # Years to process
    years = range(2011, 2018)
    success_count = 0
    fail_count = 0

    for year in years:
        transmittal_df, panel_df = None, None

        if year == 2016:
            # Handle 2016 TS data as a standalone CSV
            transmittal_csv = f"hmda_{year}_transmittal_sheet.csv"
            if os.path.exists(transmittal_csv):
                try:
                    print(f"\nProcessing transmittal sheet for 2016 as a standalone CSV...")
                    transmittal_df = pd.read_csv(
                        transmittal_csv,
                        delimiter=',',
                        engine='python',
                        on_bad_lines='warn'  # Skip problematic rows
                    )
                    transmittal_df.columns = transmittal_df.columns.str.strip().str.lower().str.replace('-', ' ')
                    
                    # Add "_ts" suffix to all transmittal sheet columns
                    transmittal_df = transmittal_df.rename(
                        columns={col: f"{col}_ts" for col in transmittal_df.columns}
                    )
                    print(f"  Loaded transmittal data from {transmittal_csv}")
                except Exception as e:
                    print(f"Error reading transmittal sheet for 2016: {e}")
                    fail_count += 1
                    continue
            else:
                print(f"Transmittal CSV file for 2016 is missing. Skipping.")
                fail_count += 1
                continue
        else:
            # Handle transmittal sheets from ZIP files for other years
            transmittal_zip = f"hmda_{year}_transmittal_sheet.zip"
            if os.path.exists(transmittal_zip):
                try:
                    with zipfile.ZipFile(transmittal_zip, 'r') as transmittal_zip_ref:
                        for file in transmittal_zip_ref.namelist():
                            if file.endswith(('.csv', '.txt')):
                                with transmittal_zip_ref.open(file) as transmittal_file:
                                    transmittal_df = pd.read_csv(
                                        transmittal_file,
                                        delimiter=',',
                                        engine='python',
                                        on_bad_lines='warn'  # Skip problematic rows
                                    )
                                    transmittal_df.columns = transmittal_df.columns.str.strip().str.lower().str.replace('-', ' ')
                                    
                                    # Add "_ts" suffix to all transmittal sheet columns
                                    transmittal_df = transmittal_df.rename(
                                        columns={col: f"{col}_ts" for col in transmittal_df.columns}
                                    )
                                    print(f"  Loaded transmittal data from {file}")
                                    break
                except Exception as e:
                    print(f"Error reading transmittal sheet for {year}: {e}")
                    fail_count += 1
                    continue
            else:
                print(f"Transmittal ZIP file for {year} is missing. Skipping.")
                fail_count += 1
                continue

        # Handle panel data from ZIP files for all years
        panel_zip = f"hmda_{year}_panel.zip"
        if os.path.exists(panel_zip):
            try:
                with zipfile.ZipFile(panel_zip, 'r') as panel_zip_ref:
                    for file in panel_zip_ref.namelist():
                        if file.endswith(('.csv', '.txt')):
                            with panel_zip_ref.open(file) as panel_file:
                                panel_df = pd.read_csv(
                                    panel_file,
                                    delimiter=',',
                                    engine='python'
                                )
                                panel_df.columns = panel_df.columns.str.strip().str.lower().str.replace('-', ' ')
                                
                                # Add "_p" suffix to all panel columns except "respondent id"
                                panel_df = panel_df.rename(
                                    columns={col: f"{col}_p" for col in panel_df.columns if col != "respondent id"}
                                )
                                print(f"  Loaded panel data from {file}")
                                break
            except Exception as e:
                print(f"Error reading panel data for {year}: {e}")
                fail_count += 1
                continue
        else:
            print(f"Panel ZIP file for {year} is missing. Skipping.")
            fail_count += 1
            continue

        # Debug column names
        print(f"Transmittal Columns ({year}): {list(transmittal_df.columns)}")
        print(f"Panel Columns ({year}): {list(panel_df.columns)}")
        
        # Check for normalized 'respondent id'
        if 'respondent id' in panel_df.columns and 'respondent id_ts' in transmittal_df.columns:
            try:
                # Merge on 'respondent id'
                merged_df = pd.merge(
                    panel_df,
                    transmittal_df,
                    left_on='respondent id',
                    right_on='respondent id_ts',
                    how='inner'
                )
                
                # Drop duplicate respondent id column from transmittal
                merged_df = merged_df.drop(columns=['respondent id_ts'])
                
                output_file = os.path.join(output_folder, f"hmda_{year}_merged.csv")
                merged_df.to_csv(output_file, index=False)
                print(f"  Merged file for {year} saved to {output_file}")
                success_count += 1
            except Exception as e:
                print(f"  Error merging data for {year}: {e}")
                fail_count += 1
        else:
            print(f"  'Respondent ID' column missing for {year}.")
            fail_count += 1

    # Final summary
    print("\nProcess Summary:")
    print(f"  Successful merges: {success_count}")
    print(f"  Failed merges: {fail_count}")

output_directory = './merged_panel_ts_files'
merge_hmda_files(output_directory)
````

````python
def summarize_merged_files(directory):
    """
    Summarizes the merged files in the given directory.
    """
    print("\nSummary of Merged Files:")
    for file in os.listdir(directory):
        if file.endswith('.csv'):
            file_path = os.path.join(directory, file)
            df = pd.read_csv(file_path)
            print(f"{file}:")
            print(f"  Total Rows: {len(df)}")
            print(f"  Unique Respondent IDs: {df['respondent id'].nunique()}")
            print("-" * 40)

output_directory = './merged_panel_ts_files'
summarize_merged_files(output_directory)
````

## Calculate the percentage of "Single Family (1-4 Units):Manufactured" for each year from 2018 to 2023 (Filtered Data)

````python
def calculate_percentage_manufactured(input_folder, years):
    """
    Calculates the percentage of 'Single Family (1-4 Units):Manufactured' in the
    'derived_dwelling_category' column for each year.

    Args:
        input_folder (str): Path to the folder containing filtered CSV files.
        years (list): List of years to process.

    Returns:
        None
    """
    for year in years:
        file_path = os.path.join(input_folder, f"{year}_filtered_lar.csv")
        
        if not os.path.exists(file_path):
            print(f"File for {year} not found. Skipping...")
            continue
        
        try:
            print(f"\nProcessing data for {year}...")
            
            # Load only the derived_dwelling_category column
            data = pd.read_csv(file_path, usecols=['derived_dwelling_category'], low_memory=False)
            
            # Count total rows and rows with 'Single Family (1-4 Units):Manufactured'
            total_rows = len(data)
            manufactured_rows = data[data['derived_dwelling_category'] == "Single Family (1-4 Units):Manufactured"].shape[0]
            
            # Calculate percentage
            percentage = (manufactured_rows / total_rows) * 100 if total_rows > 0 else 0
            print(f"  Percentage of 'Single Family (1-4 Units):Manufactured' in {year}: {percentage:.2f}%")
        
        except Exception as e:
            print(f"Error processing data for {year}: {e}")

# Example usage
input_directory = './filtered_lar_data_old'
years_to_process = list(range(2018, 2024))  # Years from 2018 to 2023
calculate_percentage_manufactured(input_directory, years_to_process)
````

# Explore variables of HMDA data files from 2011–2017 

````python
def compare_columns_across_years(input_folder, years):
    """
    Compares columns across HMDA data files from 2011 to 2017.

    Args:
        input_folder (str): Path to the folder containing ZIP files.
        years (list): List of years to process.

    Returns:
        None
    """
    columns_dict = {}  # Dictionary to store columns for each year

    for year in years:
        zip_file = os.path.join(input_folder, f"hmda_{year}_nationwide_first-lien-owner-occupied-1-4-family-records_labels.zip")
        
        if os.path.exists(zip_file):
            print(f"\nProcessing {zip_file}...")
            try:
                with zipfile.ZipFile(zip_file, 'r') as zf:
                    for file in zf.namelist():
                        if file.endswith(('.csv', '.txt')):
                            print(f"  Reading file: {file}")
                            with zf.open(file) as extracted_file:
                                # Read only the header to get columns
                                header = pd.read_csv(extracted_file, delimiter=',', nrows=0).columns.tolist()
                                columns_dict[year] = header
                                print(f"  Columns for {year} retrieved successfully.")
                            break
            except Exception as e:
                print(f"Error processing {zip_file}: {e}")
        else:
            print(f"File {zip_file} does not exist. Skipping.")

    # Compare columns across years
    print("\nComparing columns across years...")
    common_columns = set.intersection(*[set(cols) for cols in columns_dict.values()])
    all_columns = set.union(*[set(cols) for cols in columns_dict.values()])

    print(f"\nCommon columns across all years (2011–2017): {len(common_columns)} columns")
    print(common_columns)

    print(f"\nUnique columns across all years (2011–2017): {len(all_columns)} columns")
    print(all_columns)

    # Identify missing columns for each year
    for year, cols in columns_dict.items():
        missing_in_year = all_columns - set(cols)
        if missing_in_year:
            print(f"\nColumns missing in {year}: {missing_in_year}")
        else:
            print(f"\nNo columns are missing in {year}.")

# Example usage
input_directory = './'
years_to_check = list(range(2011, 2018))  # Years from 2011 to 2017
compare_columns_across_years(input_directory, years_to_check)
````

Since both the common and unique columns lists are identical and no columns are missing for any year, the datasets across these years are fully consistent.

## Excluded Columns

````python
# Columns to exclude
EXCLUDED_COLUMNS = [
    "applicant_race_1", "applicant_race_2", "applicant_race_3",
    "applicant_race_4", "applicant_race_5", "co_applicant_race_1",
    "co_applicant_race_2", "co_applicant_race_3", "co_applicant_race_4",
    "co_applicant_race_5", "applicant_race_name_1", "applicant_race_name_2",
    "applicant_race_name_3", "applicant_race_name_4", "applicant_race_name_5",
    "co_applicant_race_name_1", "co_applicant_race_name_2",
    "co_applicant_race_name_3", "co_applicant_race_name_4",
    "co_applicant_race_name_5"
]

def remove_home_improvement_and_save(input_folder, output_folder, years, output_format="csv"):
    """
    Removes entries with loan_purpose = 2 (Home Improvement) and excludes specified columns 
    from HMDA data (2011–2017) and saves the filtered data.

    Args:
        input_folder (str): Path to the folder containing ZIP files.
        output_folder (str): Path to save the filtered files.
        years (list): List of years to process.
        output_format (str): Output format, either "csv" or "excel".

    Returns:
        None
    """
    # Ensure output folder exists
    os.makedirs(output_folder, exist_ok=True)

    for year in years:
        zip_file = os.path.join(input_folder, f"hmda_{year}_nationwide_first-lien-owner-occupied-1-4-family-records_labels.zip")
        output_file = os.path.join(output_folder, f"{year}_filtered_lar.{output_format}")

        if os.path.exists(zip_file):
            print(f"\nProcessing {zip_file}...")

            try:
                with zipfile.ZipFile(zip_file, 'r') as zf:
                    for file in zf.namelist():
                        if file.endswith(('.csv', '.txt')):
                            print(f"  Reading file: {file}")
                            with zf.open(file) as extracted_file:
                                chunk_size = 100000 
                                filtered_chunks = []

                                # Read in chunks
                                for chunk in pd.read_csv(
                                    extracted_file, 
                                    delimiter=',',  
                                    chunksize=chunk_size,
                                    low_memory=False
                                ):
                                    # Filter out rows where loan_purpose = 2 (Home Improvement)
                                    if 'loan_purpose' in chunk.columns:
                                        chunk = chunk[chunk['loan_purpose'] != 2]

                                    # Drop excluded columns if they exist in the chunk
                                    chunk = chunk.drop(columns=[col for col in EXCLUDED_COLUMNS if col in chunk.columns], errors='ignore')

                                    # Append non-empty chunks
                                    if not chunk.empty:
                                        filtered_chunks.append(chunk)

                                # Concatenate all filtered chunks
                                if filtered_chunks:
                                    filtered_data = pd.concat(filtered_chunks, ignore_index=True)

                                    # Save filtered data in the specified format
                                    if output_format == "csv":
                                        filtered_data.to_csv(output_file, index=False)
                                    elif output_format == "excel":
                                        filtered_data.to_excel(output_file, index=False, engine='openpyxl')
                                    print(f"  Filtered data saved to {output_file}")
                                else:
                                    print(f"  No matching records found in {file}")
                        else:
                            print(f"  Skipped file: {file} (not a CSV or TXT)")
            except Exception as e:
                print(f"Error processing {zip_file}: {e}")
        else:
            print(f"File {zip_file} does not exist. Skipping.")

input_directory = './'
output_directory = './filtered_lar_data'
years_to_process = list(range(2011, 2018))  # Process data from 2011 to 2017
remove_home_improvement_and_save(input_directory, output_directory, years_to_process, output_format="csv")
````

# Filter the LAR data from 2018 to 2023

## **Filter Conditions**
The script filters the Loan/Application Register (LAR) data (2018–2023) based on the following conditions:

### **1. Derived Dwelling Category**
- **Field**: `derived_dwelling_category`
- **Condition**: Include only:
  - "Single Family (1-4 Units):Site-Built"

### **2. Loan Purpose**
- **Field**: `loan_purpose`
- **Condition**: Include only:
  - `1`: Home Purchase
  - `31`: Refinancing

### **3. Lien Status**
- **Field**: `lien_status`
- **Condition**: Include only:
  - `1`: Secured by a First Lien

### **4. Derived Loan Product Type**
- **Field**: `derived_loan_product_type`
- **Condition**: Include only:
  - "Conventional:First Lien"
  - "FHA:First Lien"
  - "VA:First Lien"
  - "FSA/RHS:First Lien"

### **5. Occupancy Type**
- **Field**: `occupancy_type`
- **Condition**:
  - Include: 
    - `1`: Principal Residence
  - Exclude:
    - `3`: Investment Property
    - `2`: Second Residence
   
### **6. ACTION TAKEN**
- **Field**: `action_taken`
- **Condition**:
  - Include: 
    - `1`: Loan originated
  - Exclude:
    - everything else.

### **7. Ignored Columns**
The following columns are dropped to optimize processing:
- `applicant_race_1`
- `applicant_race_2`
- `applicant_race_3`
- `applicant_race_4`
- `applicant_race_5`
- `co_applicant_race_1`
- `co_applicant_race_2`
- `co_applicant_race_3`
- `co_applicant_race_4`
- `co_applicant_race_5`

---

## **Input**
- ZIP files containing LAR data for the years 2018–2023 in the format: `YYYY_lar.zip`.
- Each ZIP file contains either `.txt` or `.csv` files with LAR data.

### **Input Example**
- `2018_lar.zip`
- `2019_lar.zip`
- `2020_lar.zip`
- ...
- `2023_lar.zip`

---

## **Output**
Filtered data files saved in the specified output directory.

### **Output Directory**
The filtered data is saved in the `./filtered_lar_data` directory.

### **Output File Names**
Each year will have its own filtered CSV file:
- `2018_filtered_lar.csv`
- `2019_filtered_lar.csv`
- `2020_filtered_lar.csv`
- ...
- `2023_filtered_lar.csv`

print columns

````python
def print_corrected_lar_columns():
    """
    Reads ZIP files with LAR data from 2018 to 2023, extracts column names correctly
    using the appropriate delimiter ('|'), and prints them for verification.
    """
    # Set the range of years to process
    years = range(2018, 2024)  # 2018 to 2023

    # Iterate through each year's ZIP file
    for year in years:
        zip_file = f"{year}_lar.zip"  # Define ZIP file name

        if os.path.exists(zip_file):
            print(f"\nProcessing {zip_file}...")
            try:
                with zipfile.ZipFile(zip_file, 'r') as zf:
                    for file in zf.namelist():
                        # Check if the file inside ZIP is a CSV or TXT
                        if file.endswith(('.csv', '.txt')):
                            print(f"  Reading file: {file}")
                            with zf.open(file) as extracted_file:
                                # Attempt to load the file and print the columns
                                try:
                                    # Specify the correct delimiter ('|') for LAR files
                                    df = pd.read_csv(extracted_file, delimiter='|', nrows=0)
                                    print(f"  Columns in {file}: {df.columns.tolist()}")
                                except Exception as e:
                                    print(f"  Error reading {file}: {e}")
                        else:
                            print(f"  Skipped file: {file} (not a CSV or TXT)")
            except Exception as e:
                print(f"Error processing {zip_file}: {e}")
        else:
            print(f"File {zip_file} does not exist. Skipping.")

print_corrected_lar_columns()
````

## filter 2018 to 2023

````python
# Filter conditions
FILTER_CONDITIONS = {
    "derived_dwelling_category": ["Single Family (1-4 Units):Site-Built"],  # Exclude Manufactured
    "loan_purpose": [1, 31],  # Home Purchase, Refinancing
    "lien_status": [1],  # First Lien
    "derived_loan_product_type": [
        "Conventional:First Lien",
        "FHA:First Lien",
        "VA:First Lien",
        "FSA/RHS:First Lien"
    ],
    "occupancy_type": [1],  # Only Principal Residence
    "action_taken": [1]  # Loan originated
}

# Define columns to exclude
EXCLUDED_COLUMNS = [
    "applicant_race_1", "applicant_race_2", "applicant_race_3",
    "applicant_race_4", "applicant_race_5", "co_applicant_race_1",
    "co_applicant_race_2", "co_applicant_race_3", "co_applicant_race_4",
    "co_applicant_race_5"
]

def efficient_filter_and_save_lar_data(input_folder, output_folder, years, output_format="csv"):
    """
    Efficiently filters and processes HMDA LAR data from ZIP archives for multiple years,
    applying filters and saving the filtered results to a specified format.

    Args:
        input_folder (str): Path to the folder containing ZIP files.
        output_folder (str): Path to save the filtered files.
        years (list): List of years to process.
        output_format (str): Output format, either "csv" or "excel".
    """
    # Ensure output folder exists
    os.makedirs(output_folder, exist_ok=True)

    for year in years:
        zip_file = os.path.join(input_folder, f"{year}_lar.zip")
        output_file = os.path.join(output_folder, f"{year}_filtered_lar.{output_format}")

        if os.path.exists(zip_file):
            print(f"\nProcessing {zip_file}...")

            try:
                with zipfile.ZipFile(zip_file, 'r') as zf:
                    for file in zf.namelist():
                        if file.endswith(('.csv', '.txt')):
                            print(f"  Reading file: {file}")
                            with zf.open(file) as extracted_file:
                                chunk_size = 100000  
                                filtered_chunks = []

                                # Read in chunks
                                for chunk in pd.read_csv(
                                    extracted_file, 
                                    delimiter='|', 
                                    chunksize=chunk_size,
                                    low_memory=False
                                ):
                                    # Exclude unwanted columns
                                    chunk = chunk[[col for col in chunk.columns if col not in EXCLUDED_COLUMNS]]

                                    # Apply filters to the chunk
                                    for col, values in FILTER_CONDITIONS.items():
                                        if col in chunk.columns:
                                            chunk = chunk[chunk[col].isin(values)]

                                    # Append non-empty chunks
                                    if not chunk.empty:
                                        filtered_chunks.append(chunk)

                                # Concatenate all filtered chunks
                                if filtered_chunks:
                                    filtered_data = pd.concat(filtered_chunks, ignore_index=True)

                                    # Save filtered data in the specified format
                                    if output_format == "csv":
                                        filtered_data.to_csv(output_file, index=False)
                                    elif output_format == "excel":
                                        filtered_data.to_excel(output_file, index=False, engine='openpyxl')
                                    print(f"  Filtered data saved to {output_file}")
                                else:
                                    print(f"  No matching records found in {file}")
                        else:
                            print(f"  Skipped file: {file} (not a CSV or TXT)")
            except Exception as e:
                print(f"Error processing {zip_file}: {e}")
        else:
            print(f"File {zip_file} does not exist. Skipping.")

input_directory = './'
output_directory = './filtered_lar_data'
years_to_process = list(range(2018, 2024))  # Process data from 2018 to 2023
efficient_filter_and_save_lar_data(input_directory, output_directory, years_to_process, output_format="csv")
````

# Verify the filtering for 2018 to 2024

````python
def verify_filtered_data(output_folder, years, period):
    """
    Verifies the filtered data for multiple years by printing unique values of the relevant columns.

    Args:
        output_folder (str): Path to the folder containing filtered CSV files.
        years (list): List of years to verify.
        period (str): Period identifier ("2011-2017" or "2018-2023").

    Returns:
        None
    """
    # Define columns to verify based on the period
    if period == "2011-2017":
        relevant_columns = [
            "property_type", "property_type_name",
            "loan_purpose", "loan_purpose_name",
            "lien_status", "lien_status_name",
            "loan_type", "loan_type_name",
            "owner_occupancy", "owner_occupancy_name",
            "action_taken", "action_taken_name"
        ]
    elif period == "2018-2023":
        relevant_columns = [
            "derived_dwelling_category", "loan_purpose", "lien_status",
            "derived_loan_product_type", "occupancy_type", "action_taken"
        ]
    else:
        print("Invalid period specified.")
        return

    for year in years:
        file_path = os.path.join(output_folder, f"{year}_filtered_lar.csv")

        if not os.path.exists(file_path):
            print(f"\nFiltered file for {year} does not exist. Skipping verification.")
            continue

        try:
            print(f"\nVerifying filtered data from {file_path}...")
            
            # Load only relevant columns to speed up processing
            data = pd.read_csv(file_path, usecols=relevant_columns, low_memory=False)
            
            # Print unique values for each relevant column
            for column in relevant_columns:
                if column in data.columns:
                    unique_values = data[column].unique()
                    print(f"  Unique values for {column} in {year}: {unique_values}")
                else:
                    print(f"  Warning: Column {column} missing in {year}")
            
            print(f"\nVerification for {year} complete.")

        except Exception as e:
            print(f"Error verifying data for {year}: {e}")

output_directory = './filtered_lar_data'

# Verify data from 2011 to 2017
years_to_verify_2011_2017 = list(range(2011, 2018))
verify_filtered_data(output_directory, years_to_verify_2011_2017, period="2011-2017")

# Verify data from 2018 to 2023
years_to_verify_2018_2023 = list(range(2018, 2024))
verify_filtered_data(output_directory, years_to_verify_2018_2023, period="2018-2023")
````
